```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
```

```python
df = pd.read_csv('../data/processed/train.csv')
display(df.head())
display(df.info())
```

```python
plt.figure(figsize=(8, 6))
ax = sns.countplot(data=df, x='target', hue='target', palette=['#333333', '#E32219'], legend=False)

plt.title('Distribution of Ferrari Finishes', fontsize=15)
plt.xlabel('Race Outcome', fontsize=12)
plt.ylabel('Number of Races', fontsize=12)
plt.xticks(ticks=[0, 1], labels=['No Podium', 'Podium'])

for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}',
                (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom', fontsize=12, fontweight='bold', color='#333333')

plt.show()
```

```python
top_grid = df[df['grid'] <= 15]

plt.figure(figsize=(12, 6))
ax = sns.barplot(data=top_grid, x='grid', y='target', color='#E32219', errorbar=None)

plt.title('Podium probability by starting position', fontsize=15)
plt.xlabel('Starting position', fontsize=12)
plt.ylabel('Probability of podium', fontsize=12)

for p in ax.patches:
    ax.annotate(f"{p.get_height()*100:.0f}%",
                (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom', fontsize=11, fontweight='bold', color='#333333')

plt.show()
```
